```python
#5A3
def checknt(n):
    if n<2:
        return False
    for i in range(2, int(n**(1/2))+1):
        if n%i==0:
            return False
    return True
    
if __name__ == '__main__':
    n = int(input())
    if checknt(n):
        print('YES')
    else:
        print('NO')
```

```python
#5A4
def hh(n):
    tonguoc = 0
    for i in range(1, n):
        if n%i==0:
            tonguoc += i
    if tonguoc == n:
        return True
    return False
if __name__ == '__main__':
    n = int(input())
    if hh(n):
        print('YES')
    else:
        print('NO')
```

```python
#5A5
def tim(lst, k):
    for i in range(len(lst)):
        if lst[i]==k:
            return i
    return -1
if __name__== '__main__':
    lst = list(map(int, input().split()))
    k = int(input())
    ketqua= tim(lst, k)
    print(ketqua)
```

```python
#5A6
def gt(n):
    ketqua = 1
    for i in range(1, n+1):
        ketqua *= i
    return ketqua
if __name__=='__main__':
    n = int(input())
    kq = gt(n)
    print(kq)
    
```

```python
#5A7
def tinh(num1, num2, operat):
    if operat == '+':
        result = num1+num2
    elif operat == '-':
        result = num1-num2
    elif operat == '*':
        result= num1*num2
    elif operat =='/':
        if num2 ==0:
            return False
        result = num1 / num2
    else:
        return 'Toán tử không hợp lệ'
    return round(result,2)
if __name__=='__main__':
    num1, num2 = map(float, input().split())
    operat = input()
    kq = tinh(num1, num2, operat)
    print(kq)
    
```

```python
#5A8
def hd(x, y):
    xor_result = x^y
    distance = bin(xor_result).count('1')
    return distance
if __name__=='__main__':
    x, y = map(int, input().split())
    kq = hd(x, y)
    print(kq)
```

```python
#5A9
def tongchuso(n):
    tong = 0
    n = abs(n)
    while n>0:
        tong += n%10
        n//=10
    return tong
if __name__ == '__main__':
    n = int(input())
    kq = tongchuso(n)
    print(kq)
```

```python
#5A10
def dc(s1, s2):
    if len(s1)!=len(s2):
        return False
    mapping = {}
    seen_value = set
    for i in range(len(s1)):
        char1 = s1[i]
        char2 = s2[i]
        if char1 in mapping:
            if mapping[char1] != char2:
                return False
            else:
                if char2 in seen_value:
                    return False
                mapping[char1] = char2
                seen_value.add(char2)
    return True
if __name__ == '__main__':
    s1 = input()
    s2 = input()
    kq = dc(s1, s2)
    print(kq)
```

```python
#5A12
def dwr(dividend, divisor):
    if divisor == 0:
        return 'Loi'
    q, r = divmod(dividend, divisor)
    return q, r
if __name__ == '__main__':
    a, b = map(int, input().split())
    cc = dwr(a, b)
    print(cc)
```
