```python
#6A4
key = list(map(str, input().split()))
d = {}
for i in range(len(key)):
    n1, n2 = map(str, key[i].split(':'))
    if n1 not in d:
        d[n1] = [n1]
    else:
        d[n1].append(n2)
print(d)
```

```python
#6A5
l = [1, -2, 0, 3, -4, 0, 0]
d = {'positives':0, 'zeros':0, 'negatives':0}
for i in l:
    if i>0:
        d['positives']+=1
    elif i==0:
        d['zeros']+=1
    else:
        d['negatives']+=1
print(d)
```

```python
#6Ak, 
d1 = eval(input())

d2 = eval(input())
res = d1.copy()
for k, v in d2.items():
    if k in res:
        res[k]+=v
    else:
        res[k]=v

sk = sorted(res.keys())
fd = {k:res[k] for k in sk}
print(fd)

```

```python
#6A10
l = list(map(int, input().split()))
k = int(input())
res= []
n = len(l)
for i in range(n):
    for j in range(i+1, n):
        if l[i]+l[j]==k:
            res.append((l[i], l[j]))
res.sort()
print(res)
```

```python
#6A12
n = [2, 3, 2, 3, 1]
soduynhat = set(n)
mc = max(n.count(x) for x in soduynhat)
sonhieunhat = [x for x in soduynhat if n.count(x)== mc]
print(min(sonhieunhat))
```
