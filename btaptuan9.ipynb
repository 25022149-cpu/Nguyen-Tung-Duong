```python
#9A1
def gop_danh_sach(ds1, ds2):
    ket_qua = []
    i = 0
    j = 0
    while i < len(ds1) and j < len(ds2):
        if ds1[i] < ds2[j]:
            ket_qua.append(ds1[i])
            i += 1
        else:
            ket_qua.append(ds2[j])
            j += 1
    ket_qua.extend(ds1[i:])
    ket_qua.extend(ds2[j:])
    return ket_qua

ds_a = list( map(int, input().split()))
ds_b =list( map(int, input().split()))
print(gop_danh_sach(ds_a, ds_b))
```

```python
#9A2
def tim_so_lon_thu_k(ma_tran, k):
    danh_sach_phang = []
    for hang in ma_tran:
        for phan_tu in hang:
            danh_sach_phang.append(phan_tu)
    
    n = len(danh_sach_phang)
    for i in range(n):
        for j in range(0, n - i - 1):
            if danh_sach_phang[j] > danh_sach_phang[j + 1]:
                danh_sach_phang[j], danh_sach_phang[j + 1] = danh_sach_phang[j + 1], danh_sach_phang[j]
    
    return danh_sach_phang[k - 1]

n, m = map(int, input().split())
ma_tran = []
for _ in range(n):
    hang = list(map(int, input().split()))
    ma_tran.append(hang)
k = int(input())

print(tim_so_lon_thu_k(ma_tran, k))
```

```python
#9A3
def hoan_doi_nho_nhat(n):
    chuoi_so = list(str(n))
    n_do_dai = len(chuoi_so)
    
    vi_tri_hoan_doi = -1
    gia_tri_thay_the = -1
    
    for i in range(n_do_dai):
        min_digit = chuoi_so[i]
        index = -1
        for j in range(n_do_dai - 1, i, -1):
            if chuoi_so[j] < min_digit:
                min_digit = chuoi_so[j]
                index = j
        
        if index != -1:
            vi_tri_hoan_doi = i
            gia_tri_thay_the = index
            break
            
    if vi_tri_hoan_doi != -1:
        chuoi_so[vi_tri_hoan_doi], chuoi_so[gia_tri_thay_the] = chuoi_so[gia_tri_thay_the], chuoi_so[vi_tri_hoan_doi]
        
    return "".join(chuoi_so)

n = input().strip()
print(hoan_doi_nho_nhat(n))
```

```python
#9A4
def kiem_tra_bien_doi(a, b):
    a_list = [c for c in a if c.isupper()]
    if "".join(a_list) == b:
        return "YES"
    return "NO"

q = int(input().strip())
for _ in range(q):
    xau_a = input().strip()
    xau_b = input().strip()
    print(kiem_tra_bien_doi(xau_a, xau_b))
```

```python
#9A5
def kiem_tra_day_con(list1, list2):
    n1 = len(list1)
    n2 = len(list2)
    
    if n1 == 0:
        return True
    if n2 == 0:
        return False
        
    for i in range(n2 - n1 + 1):
        if list2[i : i + n1] == list1:
            return True
    return False

list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))

if kiem_tra_day_con(list1, list2):
    print("True")
else:
    print("False")
```

```python
#9A6
def tim_hoan_vi_tiep_theo(n, mang):
    i = n - 2
    while i >= 0 and mang[i] >= mang[i + 1]:
        i -= 1
    
    if i == -1:
        return sorted(mang)
    
    j = n - 1
    while mang[j] <= mang[i]:
        j -= 1
        
    mang[i], mang[j] = mang[j], mang[i]
    mang[i + 1:] = reversed(mang[i + 1:])
    return mang

n = int(input())
day_so = list(map(int, input().split()))

ket_qua = tim_hoan_vi_tiep_theo(n, day_so)
print(*(ket_qua))
```

```python
#9A7
def co_the_trong_hoa(luong_hoa, k):
    dem = 0
    i = 0
    while i < len(luong_hoa):
        if luong_hoa[i] == 0:
            trung_ben_trai = (i == 0) or (luong_hoa[i - 1] == 0)
            trung_ben_phai = (i == len(luong_hoa) - 1) or (luong_hoa[i + 1] == 0)
            
            if trung_ben_trai and trung_ben_phai:
                luong_hoa[i] = 1
                dem += 1
        i += 1
    return dem >= k

du_lieu = list(map(int, input().split()))
k = int(input())

if co_the_trong_hoa(du_lieu, k):
    print("True")
else:
    print("False")
```

```python
#9A8
def kiem_tra_duong_di_lap_lai(moves):
    x, y = 0, 0
    da_di_qua = {(0, 0)}
    
    for move in moves:
        if move == 'U':
            y += 1
        elif move == 'D':
            y -= 1
        elif move == 'L':
            x -= 1
        elif move == 'R':
            x += 1
            
        if (x, y) in da_di_qua:
            return True
        da_di_qua.add((x, y))
        
    return False

chuoi_di_chuyen = input().strip()
print(kiem_tra_duong_di_lap_lai(chuoi_di_chuyen))
```

```python
#9A9
def Median(nums):
    a, b, c, d, e = nums[0], nums[1], nums[2], nums[3], nums[4]
    
    
    if a > b: a, b = b, a
    if c > d: c, d = d, c
    if a > c:
        a, c = c, a
        b, d = d, b
    
    
    if b > e: b, e = e, b
    if c > e: c, e = e, c
    if b > c: b, c = c, b
    
    return c

day_so = list(map(int, input().split()))
print(Median(day_so))
```

```python
#9A10
def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()
    
    def kiem_tra(r):
        i = 0
        for h in houses:
            while i < len(heaters) and heaters[i] + r < h:
                i += 1
            if i < len(heaters) and heaters[i] - r <= h <= heaters[i] + r:
                continue
            else:
                return False
        return True

    trai, phai = 0, 10**9
    ket_qua = phai
    while trai <= phai:
        mid = (trai + phai) // 2
        if kiem_tra(mid):
            ket_qua = mid
            phai = mid - 1
        else:
            trai = mid + 1
    return ket_qua

houses = list(map(int, input().split()))
heaters = list(map(int, input().split()))
print(findRadius(houses, heaters))
```

```python
#9A11
def f(n):
    if n == 0 or n == 1:
        return 1
    return n * f(n - 1)

n = int(input())
print(f(n))
```

```python
#9A12
def sum_to_n(n):
    if n == 1:
        return 1
    return n + sum_to_n(n - 1)

n = int(input())
print(sum_to_n(n))
```

```python
#9A13
def power(base, exp):
    if exp == 0:
        return 1
    if exp == 1:
        return base
    
    nua_mu = power(base, exp // 2)
    
    if exp % 2 == 0:
        return nua_mu * nua_mu
    else:
        return base * nua_mu * nua_mu

base = int(input())
exp = int(input())
print(power(base, exp))
```

```python
#9A14
def tim_so_xuat_hien_nhieu_hon(day_so, x):
    ket_qua = []
    da_xu_ly = []
    for num in day_so:
        if num not in da_xu_ly:
            dem = 0
            for i in day_so:
                if i == num:
                    dem += 1
            if dem > x:
                ket_qua.append(num)
            da_xu_ly.append(num)
    
    if not ket_qua:
        return "INVALID"
    return " ".join(map(str, ket_qua))

day_so = list(map(int, input().split()))
x = int(input())
print(tim_so_xuat_hien_nhieu_hon(day_so, x))
```

```python
#9A15
def kiem_tra_day_con(day_con, day_me):
    n, m = len(day_con), len(day_me)
    if n == 0:
        return True
    
    i = 0
    j = 0
    while i < n and j < m:
        if day_con[i] == day_me[j]:
            i += 1
        j += 1
    
    return i == n

day_con = list(map(int, input().split()))
day_me = list(map(int, input().split()))

if kiem_tra_day_con(day_con, day_me):
    print("True")
else:
    print("False")
```
