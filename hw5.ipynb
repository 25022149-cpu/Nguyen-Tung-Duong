```python
#5A24
def ac(x1, y1, x2, y2, x3, y3):
    val = (x2 - x1) * (y3 - y1) - (y2 - y1) * (x3 - x1)
    if val == 0:
        return True
    else:
        return False
if __name__ == "__main__":
    x1, y1, x2, y2, x3, y3= map(int, input().split())
    if ac(x1, y1, x2, y2, x3, y3):
        print('True')
    else:
        print('False')
```

```python
#5A25

def piq(quad, p):
    n = len(quad)
    xp, yp = p
    sides = []

    for i in range(n):
        x1, y1 = quad[i]
        x2, y2 = quad[(i + 1) % n]
        val = (x2 - x1) * (yp - y1) - (y2 - y1) * (xp - x1)
        sides.append(val)
    is_all_non_negative = all(s >= 0 for s in sides)
    is_all_non_positive = all(s <= 0 for s in sides)

    return is_all_non_negative or is_all_non_positive

if __name__ == "__main__":
    quad_sample = [(0, 0), (4, 0), (4, 3), (0, 3)]
    p_sample = (2, 1)
    
    print(f"Tứ giác: {quad_sample}")
    print(f"Điểm cần kiểm tra: {p_sample}")
    
```

```python
#5A27
def count_non_zero_digits(n_list):
    total_count = 0
    
    for num in n_list:
        
        temp = abs(num)
        
        
        if temp == 0:
            continue
            
        while temp > 0:
            digit = temp % 10
            if digit != 0:    
                total_count += 1
            temp //= 10       
            
    return total_count

if __name__ == "__main__":
    
    test_list1 = [102, 300, 45] 
    
    print(f" Với danh sách {test_list1}")
    print(f"Số lượng chữ số khác 0 là: {count_non_zero_digits(test_list1)}")
```

```python
#28
import math

def calculate_angle_abc(x1, y1, x2, y2, x3, y3):
   
    ax, ay = x1 - x2, y1 - y2  
    bx, by = x3 - x2, y3 - y2  
    dot_product = ax * bx + ay * by
    
    mag_a = math.sqrt(ax**2 + ay**2)
    mag_b = math.sqrt(bx**2 + by**2)
    
    
    if mag_a == 0 or mag_b == 0:
        return 0
    
    
    cos_theta = dot_product / (mag_a * mag_b)
    
    
    if cos_theta > 1:
        cos_theta = 1
    elif cos_theta < -1:
        cos_theta = -1
        
    
    angle_rad = math.acos(cos_theta)
    angle_deg = math.degrees(angle_rad)
    
    return angle_deg
if __name__ == "__main__":
    print("--- Kiểm tra hàm tính góc ABC ---")
    
    angle1 = calculate_angle_abc(0, 1, 0, 0, 1, 0)
    print(f" A(0,1), B(0,0), C(1,0) => Góc ABC = {angle1:.1f} độ")
    
```

```python
#29
def get_max_min(a, b, c):
    
    if a >= b and a >= c:
        max_val = a
    elif b >= a and b >= c:
        max_val = b
    else:
        max_val = c
        
    
    if a <= b and a <= c:
        min_val = a
    elif b <= a and b <= c:
        min_val = b
    else:
        min_val = c
        
    return [max_val, min_val]


if __name__ == "__main__":
    print("--- Kiểm tra hàm get_max_min ---")
    
    
    num1, num2, num3 = 15, 42, 23
    result1 = get_max_min(num1, num2, num3)
    print(f" Ba số ({num1}, {num2}, {num3})")
    print(f"=> Kết quả [max, min]: {result1}")
    
    
    
```

```python
#30
import math

def combinations(n, k):
    if k < 0 or k > n:
        return 0
    
    
    if k == 0 or k == n:
        return 1
    
    result = math.factorial(n) // (math.factorial(k) * math.factorial(n - k))
    
    return result


if __name__ == "__main__":
    print("--- Kiểm tra hàm combinations C(n, k) ---")
    
    
    n1, k1 = 5, 2
    print(f" C({n1}, {k1}) = {combinations(n1, k1)}")
    
    
```

```python
#31
def sum_squares(n):
    total = 0
    
    if n <= 0:
        return 0
    
    
    for i in range(1, n + 1):
        total += i * i
        
    return total


if __name__ == "__main__":
    print("--- Kiểm tra hàm sum_squares ---")
    
    
    val1 = 3
    print(f" Tổng bình phương từ 1 đến {val1} là: {sum_squares(val1)}")
    
    
```

```python
#32
def calculate_sum_from_string(expression):
    if not expression:
        return 0
    
    try:
        result = eval(expression)
        return result
    except:
        
        return "Biểu thức không hợp lệ"

if __name__ == "__main__":
    print("--- Kiểm tra hàm calculate_sum_from_string ---")
    expr1 = "15 + 25"
    print(f" Biểu thức '{expr1}' => Kết quả: {calculate_sum_from_string(expr1)}")
    
    
```

```python
#33
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    
    if car_type == "4seats":
        open_price = 10000
        price_per_km = 12000
    else:
        open_price = 12000
        price_per_km = 15000
        
    fare = open_price + (km * price_per_km) + (wait_minutes * 500)
    
    
    if start_hour >= 22 or start_hour <= 5:
        
        fare = fare * 1.15
        
    return fare


if __name__ == "__main__":
    print("--- Kiểm tra hàm tính tiền taxi smart_taxi_fare ---")
    
    
    fare1 = smart_taxi_fare("4seats", 10, 10, 0)
    print(f"(4 chỗ, 10h sáng, 10km): {fare1:,} VNĐ")
    
   
```

```python
#34
def has_unique_digits(num):
    s = str(num)
    n = len(s)
    
    
    for i in range(n):
        for j in range(i + 1, n):
            
            if s[i] == s[j]:
                return False
    return True

def sum_unique_digits_numbers(n):
   
    total = 0
    
   
    if n <= 0:
        return 0
        
    for x in range(1, n + 1):
        
        if has_unique_digits(x):
            total += x
            
    return total


if __name__ == "__main__":
    print("--- Kiểm tra hàm sum_unique_digits_numbers ---")
    
    
    val1 = 12
    print(f" n = {val1}")
    print(f"=> Tổng các số có chữ số khác nhau: {sum_unique_digits_numbers(val1)}")
    
   
```
