```python
#9A16
def sinh_chuoi_nhi_phan(so_luong_0, so_luong_1, chuoi_hien_tai):
    if so_luong_0 == 0 and so_luong_1 == 0:
        print(chuoi_hien_tai)
        return

    if so_luong_0 > 0:
        sinh_chuoi_nhi_phan(so_luong_0 - 1, so_luong_1, chuoi_hien_tai + "0")

    if so_luong_1 > 0:
        sinh_chuoi_nhi_phan(so_luong_0, so_luong_1 - 1, chuoi_hien_tai + "1")


du_lieu_vao = input()
thanh_phan = du_lieu_vao.split()
n = int(input())
m = int(input())

sinh_chuoi_nhi_phan(n, m, "")
```

```python
#9a17
def tim_day_con_tang_dai_nhat(danh_sach):
    so_phan_tu = len(danh_sach)
    if so_phan_tu == 0:
        return 0

    luu_tru = [1] * so_phan_tu

    for i in range(1, so_phan_tu):
        for j in range(0, i):
            if danh_sach[i] > danh_sach[j]:
                if luu_tru[i] < luu_tru[j] + 1:
                    luu_tru[i] = luu_tru[j] + 1

    do_dai_lon_nhat = luu_tru[0]
    for gia_tri in luu_tru:
        if gia_tri > do_dai_lon_nhat:
            do_dai_lon_nhat = gia_tri

    return do_dai_lon_nhat


dong_vao = input()
chuoi_so = dong_vao.split()
day_so = []
for chuoi in chuoi_so:
    day_so.append(int(chuoi))

ket_qua = tim_day_con_tang_dai_nhat(day_so)
print(ket_qua)
```

```python
#9A20
chuoi_vao = input()
chuoi_thuong = ""
for ky_tu in chuoi_vao:
    if "A" <= ky_tu <= "Z":
        chuoi_thuong += chr(ord(ky_tu) + 32)
    else:
        chuoi_thuong += ky_tu
danh_sach_tu = []
tu_hien_tai = ""
for ky_tu in chuoi_thuong:
    if ky_tu != " " and ky_tu != "\t" and ky_tu != "\n":
        tu_hien_tai += ky_tu
    else:
        if tu_hien_tai != "":
            danh_sach_tu.append(tu_hien_tai)
            tu_hien_tai = ""
if tu_hien_tai != "":
    danh_sach_tu.append(tu_hien_tai)
chuoi_ket_qua = ""
index_tu = 0
for tu in danh_sach_tu:
    tu_chuan_hoa = ""
    index_ky_tu = 0
    for ky_tu in tu:
        if index_ky_tu == 0:
            if "a" <= ky_tu <= "z":
                tu_chuan_hoa += chr(ord(ky_tu) - 32)
            else:
                tu_chuan_hoa += ky_tu
        else:
            tu_chuan_hoa += ky_tu
        index_ky_tu += 1
    if index_tu == 0:
        chuoi_ket_qua += tu_chuan_hoa
    else:
        chuoi_ket_qua += " " + tu_chuan_hoa
    index_tu += 1
print(chuoi_ket_qua)
```

```python
#9A21
def phan_hoach(danh_sach, chi_so_dau, chi_so_cuoi):
    chot = danh_sach[chi_so_cuoi]
    i = chi_so_dau
    for j in range(chi_so_dau, chi_so_cuoi):
        if danh_sach[j] <= chot:
            danh_sach[i], danh_sach[j] = danh_sach[j], danh_sach[i]
            i += 1
    danh_sach[i], danh_sach[chi_so_cuoi] = danh_sach[chi_so_cuoi], danh_sach[i]
    return i


def lua_chon_nhanh(danh_sach, chi_so_dau, chi_so_cuoi, vi_tri_muc_tieu):
    if chi_so_dau == chi_so_cuoi:
        return danh_sach[chi_so_dau]

    vi_tri_chot = phan_hoach(danh_sach, chi_so_dau, chi_so_cuoi)

    if vi_tri_muc_tieu == vi_tri_chot:
        return danh_sach[vi_tri_muc_tieu]
    elif vi_tri_muc_tieu < vi_tri_chot:
        return lua_chon_nhanh(danh_sach, chi_so_dau, vi_tri_chot - 1, vi_tri_muc_tieu)
    else:
        return lua_chon_nhanh(danh_sach, vi_tri_chot + 1, chi_so_cuoi, vi_tri_muc_tieu)


dong_nhap_day = input()
chuoi_cac_so = dong_nhap_day.split()
day_so = []
for chuoi in chuoi_cac_so:
    day_so.append(int(chuoi))

k = int(input())

so_luong = len(day_so)
muc_tieu = so_luong - k

ket_qua = lua_chon_nhanh(day_so, 0, so_luong - 1, muc_tieu)
print(ket_qua)
```

```python
#9A22
def kiem_tra_anagram(chuoi_mot, chuoi_hai):
    if len(chuoi_mot) != len(chuoi_hai):
        return False

    mang_dem = [0] * 256

    for ky_tu in chuoi_mot:
        ma_ascii = ord(ky_tu)
        mang_dem[ma_ascii] += 1

    for ky_tu in chuoi_hai:
        ma_ascii = ord(ky_tu)
        mang_dem[ma_ascii] -= 1

    for so_lan in mang_dem:
        if so_lan != 0:
            return False

    return True


chuoi_thu_nhat = input()
chuoi_thu_hai = input()

if kiem_tra_anagram(chuoi_thu_nhat, chuoi_thu_hai):
    print("True")
else:
    print("False")
```

```python
#9A23
chuoi_so = input()
k = int(input())

so_luong_can_xoa = len(chuoi_so) - k
ngan_xep = []

for chu_so in chuoi_so:
    while so_luong_can_xoa > 0 and len(ngan_xep) > 0 and ngan_xep[-1] < chu_so:
        ngan_xep.pop()
        so_luong_can_xoa -= 1
    ngan_xep.append(chu_so)

while so_luong_can_xoa > 0:
    ngan_xep.pop()
    so_luong_can_xoa -= 1

ket_qua = ""
for i in range(k):
    ket_qua += ngan_xep[i]

print(ket_qua)
```
