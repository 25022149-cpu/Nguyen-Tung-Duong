```python
#7A16
def md(arr):
    if not arr:
        return 0
    count = {}
    for i in arr:
        if i in count:
            count[i]+=1
        else:
            count[i]=1
    canxoa = 0
    for x, cou in count.items():
        if cou >= x:
            canxoa +=min(cou -x, cou)
        else:
            canxoa +=cou
    return canxoa
arr = list(map(int, input().split()))
print(md(arr))
```

```python
#7A17
def f(arr):
    n = len(arr)
    if n ==0:
        return None
    c = None
    count = 0
    for num in arr:
        if count ==0:
            c = num
            count = 1
        elif num == c:
            count +=1
        else:
            count -=1
    so = 0
    for num in arr:
        if num == c:
            so+=1
    if so >n//2:
        return c
    else:
        return None
arr = list(map(int, input().split()))
print(f(arr))

```

```python
#7A18
def so(arr):
    c = []
    l= []
    for i in arr:
        if i %2==0:
            c.append(i)
        else:
            l.append(i)
    c.sort()
    l.sort(reverse=True)
    return c +l
arr = list(map(int, input().split()))
print(so(arr))
            
```

```python
#7A19
def merge(arr1, arr2):
    if len(arr1)>1 and arr1[0]>arr1[-1]:
        arr1 = arr1[::-1]
    if len(arr2)>1 and arr2[0]>arr2[-1]:
        arr2 = arr2[::-1]
    kq = []
    i = 0
    j= 0
    while i <len(arr1) and j < len(arr2):
        if arr1[i]< arr2[j]:
            kq.append(arr1[i])
            i+=1
        else:
            kq.append(arr2[j])
            j+=1
    while i < len(arr1):
        kq.append(arr1[i])
        i+=1
    while j<len(arr2):
        kq.append(arr2[j])
        j+=1
    return kq
arr1 = list(map(int, input().split()))
arr2 = list(map(int, input().split()))
print(merge(arr1, arr2))
```

```python
#7A22
def ten_pho_bien(ds_ho_ten):
    ds_ten = []
    for ho_ten in ds_ho_ten:
        cum_tu = ho_ten.strip().split()
        if cum_tu:
            ds_ten.append(cum_tu[-1])
            
    if not ds_ten:
        return None

    dem = {}
    for ten in ds_ten:
        if ten in dem:
            dem[ten] += 1
        else:
            dem[ten] = 1

    kq = None
    max_sl = -1

    for ten, sl in dem.items():
        if sl > max_sl:
            max_sl = sl
            kq = ten
        elif sl == max_sl:
            if ten < kq:
                kq = ten

    return kq

ds = input().split(",")
print(ten_pho_bien(ds))
```

```python
#7A23
def dem_ten(xau):
    sl = 0
    tu_hien_tai = ""
    
    for kt in xau:
        if kt.isalpha():
            tu_hien_tai += kt
        else:
            if len(tu_hien_tai) >= 2 and tu_hien_tai[0].isupper() and tu_hien_tai[1:].islower():
                sl += 1
            tu_hien_tai = ""
            
    if len(tu_hien_tai) >= 2 and tu_hien_tai[0].isupper() and tu_hien_tai[1:].islower():
        sl += 1
        
    return sl

xau_vao = input()
print(dem_ten(xau_vao))
```

```python
#7A24
def xoa_trung_lien_ke(xau):
    tu_hoc = xau.split()
    if not tu_hoc:
        return ""
    
    kq = [tu_hoc[0]]
    for i in range(1, len(tu_hoc)):
        if tu_hoc[i] != tu_hoc[i-1]:
            kq.append(tu_hoc[i])
            
    return " ".join(kq)

xau_vao = input()
print(xoa_trung_lien_ke(xau_vao))
```

```python
#7A25
def so_lon_nhat(xau_so, k):
    kq = []
    so_chu_so_can_xoa = len(xau_so) - k
    
    for chu_so in xau_so:
        while so_chu_so_can_xoa > 0 and kq and kq[-1] < chu_so:
            kq.pop()
            so_chu_so_can_xoa -= 1
        kq.append(chu_so)
        
    return "".join(kq[:k])

xau_vao = input()
k_vao = int(input())
print(so_lon_nhat(xau_vao, k_vao))
```

```python
#7A26
def tinh_tong_trung_binh(xau):
    ds_so = []
    so_hien_tai = ""
    
    for kt in xau:
        if kt.isdigit():
            so_hien_tai += kt
        else:
            if so_hien_tai:
                ds_so.append(int(so_hien_tai))
                so_hien_tai = ""
                
    if so_hien_tai:
        ds_so.append(int(so_hien_tai))
        
    if not ds_so:
        return 0, 0
        
    tong = sum(ds_so)
    tb = tong / len(ds_so)
    return tong, tb

xau_vao = input()
tong_so, trung_binh = tinh_tong_trung_binh(xau_vao)
print(tong_so)
print(trung_binh)
```

```python
#7A27
def kiem_tra_ngoac(xau):
    ngan_xep = []
    cap_ngoac = {')': '(', '}': '{', ']': '['}
    
    for kt in xau:
        if kt in cap_ngoac.values():
            ngan_xep.append(kt)
        elif kt in cap_ngoac.keys():
            if not ngan_xep or ngan_xep[-1] != cap_ngoac[kt]:
                return False
            ngan_xep.pop()
            
    return len(ngan_xep) == 0

xau_vao = input()
if kiem_tra_ngoac(xau_vao):
    print("Hop le")
else:
    print("Khong hop le")
```

```python
#7A28
def check_nguyen_to(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def day_hop_so_dai_nhat(n):
    max_day = []
    day_hien_tai = []
    
    for i in range(2, n + 1):
        if not check_nguyen_to(i):
            day_hien_tai.append(i)
        else:
            if len(day_hien_tai) > len(max_day):
                max_day = day_hien_tai
            day_hien_tai = []
            
    if len(day_hien_tai) > len(max_day):
        max_day = day_hien_tai
        
    return max_day

n_vao = int(input())
kq = day_hop_so_dai_nhat(n_vao)
print(kq)
```

```python
#7A29
def day_tang_lien_tuc(ds):
    if not ds:
        return []
        
    max_day = [ds[0]]
    day_hien_tai = [ds[0]]
    
    for i in range(1, len(ds)):
        if ds[i] > ds[i-1]:
            day_hien_tai.append(ds[i])
        else:
            if len(day_hien_tai) > len(max_day):
                max_day = day_hien_tai
            day_hien_tai = [ds[i]]
            
    if len(day_hien_tai) > len(max_day):
        max_day = day_hien_tai
        
    return max_day

ds_vao = [int(x) for x in input().split()]
print(day_tang_lien_tuc(ds_vao))
```

```python
#7A30
def day_tang_khong_lien_tuc(ds):
    if not ds:
        return []
        
    n = len(ds)
    f = [1] * n
    truoc = [-1] * n
    
    for i in range(1, n):
        for j in range(i):
            if ds[i] > ds[j] and f[j] + 1 > f[i]:
                f[i] = f[j] + 1
                truoc[i] = j
                
    max_f = max(f)
    cs_max = f.index(max_f)
    
    kq = []
    while cs_max != -1:
        kq.append(ds[cs_max])
        cs_max = truoc[cs_max]
        
    kq.reverse()
    return kq

ds_vao = [int(x) for x in input().split()]
print(day_tang_khong_lien_tuc(ds_vao))
```

```python
#7A31
def so_nho_nhat(xau):
    n = len(xau)
    a = list(xau)
    
    vi_tri_doi_1 = -1
    vi_tri_doi_2 = -1
    
    for i in range(n):
        cs_min = a[i]
        vi_tri_min = -1
        
        for j in range(n - 1, i, -1):
            if a[j] < cs_min:
                if i == 0 and a[j] == '0':
                    continue
                cs_min = a[j]
                vi_tri_min = j
                
        if vi_tri_min != -1:
            vi_tri_doi_1 = i
            vi_tri_doi_2 = vi_tri_min
            break
            
    if vi_tri_doi_1 != -1:
        a[vi_tri_doi_1], a[vi_tri_doi_2] = a[vi_tri_doi_2], a[vi_tri_doi_1]
        
    return "".join(a)

xau_vao = input()
print(so_nho_nhat(xau_vao))
```
