```python
#10A3
try:
    chuoi_nhap = input()
    so_nguyen = int(chuoi_nhap)

    tap_hop_da_duyet = []
    la_so_hanh_phuc = False

    while so_nguyen not in tap_hop_da_duyet:
        tap_hop_da_duyet.append(so_nguyen)
        tong_binh_phuong = 0
        chuoi_tam = str(so_nguyen)
        for ky_tu in chuoi_tam:
            tong_binh_phuong += int(ky_tu) ** 2
        so_nguyen = tong_binh_phuong
        if so_nguyen == 1:
            la_so_hanh_phuc = True
            break

    print(la_so_hanh_phuc)

except ValueError:
    print("ValueError")
```

```python
#10A10
so_n = int(input())
dong_mang = input()
chuoi_cac_so = dong_mang.split()

if len(chuoi_cac_so) != so_n:
    print("Mang khong hop le")
else:
    mang_so = []
    hop_le = True
    tong_cac_phan_tu = 0

    for chuoi in chuoi_cac_so:
        gia_tri = int(chuoi)
        if gia_tri in mang_so:
            hop_le = False
            break
        mang_so.append(gia_tri)
        tong_cac_phan_tu += gia_tri

    if hop_le:
        print(tong_cac_phan_tu)
    else:
        print("Mang khong hop le")
```
