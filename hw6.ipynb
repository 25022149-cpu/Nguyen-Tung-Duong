```python
#6A19
def f(lst):
    if not lst:  
        return -1
    m = max(lst)
    return lst.index(m)
nums = list(map(int, input().split()))
print("Vị trí max đầu tiên:", f(nums))
```

```python
#6A22
def tim_vi_tri_max(danh_sach):
    if not danh_sach:
        return None
    
    so_lon_nhat = max(danh_sach)
    
    vi_tri = danh_sach.index(so_lon_nhat)
    
    return vi_tri
danh_sach = list(map(int, input().split()))
print("Vị trí của số lớn nhất là:", tim_vi_tri_max(danh_sach)) 
```

```python
#6A23
def xoa_so_nho_hon_x(mang, x):
    ket_qua = []
    
    for so in mang:
        if so >= x:
            ket_qua.append(so) 
    return ket_qua
my_list = list(map(int, input().split()))
x = int(input())
print(xoa_so_nho_hon_x(my_list, x)) 
```

```python
#6A24
def day_so_0_ve_dau(mang):
    hop_so_0 = []
    hop_so_khac = []
    for so in mang:
        if so == 0:
            hop_so_0.append(so)    
        else:
            hop_so_khac.append(so) 
    ket_qua = hop_so_0 + hop_so_khac
    
    return ket_qua
mang = list(map(int, input().split()))
print("Kết quả:", day_so_0_ve_dau(mang))
```

```python
#6A25
def f(lst):
    m = []
    for i in range(10):
        if i not in lst:
            m.append(i)
    return m
lst = list(map(int, input().split()))
print("Các số còn thiếu:", f(lst))
```

```python
#6A26
def cd(lst):
    tong = 0
    for i in range(1, len(lst)-1):
        if lst[i]>lst[i-1] and lst[i]>lst[i+1]:
            tong+=lst[i]
    return tong
lst = list(map(int, input().split()))
print(cd(lst))
```

```python
#6A27
def tim_vi_tri_nho_hon_x(mang, x):
    vi_tri_tim_duoc = []
    for i, so in enumerate(mang):
        if so < x:
            vi_tri_tim_duoc.append(i)
    if len(vi_tri_tim_duoc) == 0:
        print("-1")
    else:
        print(*(vi_tri_tim_duoc))

day_so = list(map(int, input().split()))
x_can_tim = int(input())
tim_vi_tri_nho_hon_x(day_so, x_can_tim)
```

```python
#6A28
def kiem_tra_day_so(mang):
    if len(mang) < 2:
        return "Vô danh"
    la_tang = True
    la_giam = True
    for i in range(len(mang) - 1):
        if not (mang[i] < mang[i+1]):
            la_tang = False
        if not (mang[i] > mang[i+1]):
            la_giam = False
    if la_tang:
        return "Tăng thực sự"
    elif la_giam:
        return "Giảm thực sự"
    else:
        return "Vô danh"
mang = list(map(int, input().split()))
print( kiem_tra_day_so(mang))
```
