```python
#7A1
def ls(arr, left, right, target):
    while left<=right:
        mid = (left+right)//2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid +1
        else:
            right = mid -1
    return -1
arr = list(map(int, input().split()))
target = int(input())
print(ls(arr, 0, len(arr)-1, target))
```

```python
#7A3
def bs(arr):
    n = len(arr)
    for i in range(n):
        mini = i
        for j in range(i+1, n):
            if arr[j]<arr[mini]:
                mini = j
        arr[i], arr[mini] = arr[mini], arr[i]
    return arr
arr = list(map(int, input().split()))
print(bs(arr))
```

```python
#7A4
def dem(arr, x):
    n= len(arr)
    count=0
    
    for i in range(n):
    
        for j in range(i+1, n):
            if arr[i] + arr[j] == x:
                count += 1
    return count
arr = list(map(int, input().split()))
x = int(input())
print(dem(arr, x))
```

```python
#7A7
def ddln(arr):
    n = len(arr)
    k = [1]*n
    for i in range(1, n):
        for j in range(0, i):
            if arr[i]>arr[j]:
                k[i] = max(k[i], k[j]+1)
    return k[i]
arr = list(map(int, input().split()))
print(ddln(arr))
```

```python
#7A8
def tg(inte, que):
    inte.sort(key = lambda x: (x[1]-x[0]+1, x[0]))
    res = []
    for q in que:
        m = False
        for l, r in inte:
            if l <= q<= r:
                res.append(r-l+1)
                m = True
                break
        if not m:
            res.append(-1)
    return res
inte = eval(input())
que = eval(input())
print(tg(inte, que))
```

```python
#7A9
def bp(arr):
    n = len(arr)
    res = [0]*n
    for i in range(n):
        count = 0
        for j in range(i+1, n):
            if arr[j]<arr[i]:
                count+=1
        res[i] = count
    return res
arr = eval(input())
print(bp(arr))
```

```python
#7A12
def sxnhanh(pa):
    n = len(pa)
    for i in range(n):
        for j in range(0, n-i-1):
            if pa[j][0]> pa[j+1][0]:
                pa[j], pa[j+1]= pa[j+1], pa[j]
    kq = ''
    for s in pa:
        kq += s[1] + ' '
    return kq.strip()
while True:
    l1 = input().strip()
    if l1 == '':
        break
    l2 = input().strip()
    if l2 == '':
        break
    l3 = input().strip()
    if l3 == '':
        break
    l4 = input().strip()
    if l4 == '':
        break
    ds = [
        [l1, 'A'],
        [l2, 'B'],
        [l3, 'C'],
        [l4, 'D']

    ]
    print(sxnhanh(ds))
```

```python
#7A13
def dhh(u):
    count = {}
    for num in u:
        if num in count:
            count[num]+=1
        else:
            count[num]=1
    dainhat = 0
    for x in count:
        if x+1 in count:
            c = count[x]+count[x+1]
            if c> dainhat:
                dainhat = c
    return dainhat
while True:
    l = input().strip()
    if l =='':
        break
    if ')' in l:
        l = l.split(')')[1].strip()
    u = list(map(int, l.split()))
    print(dhh(u))        
```

```python
#7A14
def tim(u):
    kq =[]
    for i in range(len(u)):
        if u[i] == 7:
            kq.append(str(i))
    if len(kq)==0:
        return "Not found"
    kq.reverse()
    return " ".join(kq)
while True:
    l1 = input().strip()
    if l1 == "":
        break
    l2 = input().strip()
    u = list(map(int, l2.split()))
    print(tim(u))
```
