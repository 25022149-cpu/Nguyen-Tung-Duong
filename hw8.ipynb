```python
#8A16
class Hinh:
    def dien_tich(self):
        return 0


class HinhChuNhat(Hinh):
    def __init__(self, chieu_rong, chieu_cao):
        self.chieu_rong = chieu_rong
        self.chieu_cao = chieu_cao

    def dien_tich(self):
        return self.chieu_rong * self.chieu_cao


class HinhTron(Hinh):
    def __init__(self, ban_kinh):
        self.ban_kinh = ban_kinh

    def dien_tich(self):
        return 3.14 * self.ban_kinh * self.ban_kinh


loai_hinh = input().strip()

if loai_hinh == "RECTANGLE":
    rong, cao = map(float, input().split())
    hinh = HinhChuNhat(rong, cao)
else:
    ban_kinh = float(input())
    hinh = HinhTron(ban_kinh)

print(f"{hinh.dien_tich():.2f}")
```

```python
#8A17
class PhuongTien:
    def __init__(self, nhan_hieu, toc_do):
        self.nhan_hieu = nhan_hieu
        self.toc_do = toc_do
    def thong_tin(self):
        return f"{self.nhan_hieu}-{self.toc_do}"

class OTo(PhuongTien):
    def __init__(self, nhan_hieu, toc_do, so_cho):
        super().__init__(nhan_hieu, toc_do)
        self.so_cho = so_cho
    def thong_tin(self):
        return f"{self.nhan_hieu}-{self.toc_do}-{self.so_cho}"

dong1 = input().strip()
dong2 = input().split()

if dong1 == "Vehicle":
    doi_tuong = PhuongTien(dong2[0], dong2[1])
elif dong1 == "Car":
    doi_tuong = OTo(dong2[0], dong2[1], dong2[2])

print(doi_tuong.thong_tin())
```

```python
#8A18
class DaGiac:
    def tinh_chu_vi(self):
        pass

class TamGiac(DaGiac):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def tinh_chu_vi(self):
        return self.a + self.b + self.c

a, b, c = map(int, input().split())
doi_tuong = TamGiac(a, b, c)
print(doi_tuong.tinh_chu_vi())
```

```python
#8A19
class Nguoi:
    def __init__(self, ten, tuoi):
        self.ten = ten
        self.tuoi = tuoi
    def mo_ta(self):
        return f"{self.ten} {self.tuoi}"

class SinhVien(Nguoi):
    def __init__(self, ten, tuoi, truong):
        super().__init__(ten, tuoi)
        self.truong = truong
    def mo_ta(self):
        return f"{self.ten} {self.tuoi} {self.truong}"

loai = input().strip()
du_lieu = input().split()

if loai == "Person":
    doi_tuong = Nguoi(du_lieu[0], du_lieu[1])
elif loai == "Student":
    doi_tuong = SinhVien(du_lieu[0], du_lieu[1], du_lieu[2])

print(doi_tuong.mo_ta())
```

```python
#8A20
class TaiKhoan:
    def __init__(self, so_du):
        self.so_du = so_du
    def nap_tien(self, x):
        self.so_du += x
    def lay_so_du(self):
        return self.so_du

class TaiKhoanTietKiem(TaiKhoan):
    def __init__(self, so_du, lai_suat):
        super().__init__(so_du)
        self.lai_suat = lai_suat
    def cong_lai(self):
        self.so_du += self.so_du * self.lai_suat

du_lieu = input().split()
so_du = float(du_lieu[0])
lai_suat = float(du_lieu[1])

doi_tuong = TaiKhoanTietKiem(so_du, lai_suat)
doi_tuong.cong_lai()

print(f"{doi_tuong.lay_so_du():.2f}")
```

```python
#8A21
class MayIn:
    def in_gia_tri(self, x):
        pass

class MayInNhiPhan(MayIn):
    def in_gia_tri(self, x):
        return bin(x)[2:]

n = int(input().strip())
doi_tuong = MayInNhiPhan()
print(doi_tuong.in_gia_tri(n))
```

```python
#8A22
class MaTran2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def vet(self):
        return self.a + self.d
    def dinh_thuc(self):
        return self.a * self.d - self.b * self.c

a, b, c, d = map(int, input().split())
doi_tuong = MaTran2x2(a, b, c, d)
print(f"{doi_tuong.vet()} {doi_tuong.dinh_thuc()}")
```

```python
#8A23
class Diem:
    def __init__(self, x, y):
        self.x = x
        self.y = y

class KhoangCach:
    def tinh_khoang_cach(self, d1, d2):
        return ((d2.x - d1.x) ** 2 + (d2.y - d1.y) ** 2) ** 0.5

x1, y1 = map(float, input().split())
x2, y2 = map(float, input().split())

d1 = Diem(x1, y1)
d2 = Diem(x2, y2)
tinh = KhoangCach()

print(f"{tinh.tinh_khoang_cach(d1, d2):.2f}")
```

```python
#8A24
class HinhChuNhat:
    def __init__(self, chieu_dai, chieu_rong):
        self.chieu_dai = chieu_dai
        self.chieu_rong = chieu_rong
    def tinh_chu_vi(self):
        return (self.chieu_dai + self.chieu_rong) * 2
    def tinh_dien_tich(self):
        return self.chieu_dai * self.chieu_rong

dai, rong = map(float, input().split())
doi_tuong = HinhChuNhat(dai, rong)

print(f"{doi_tuong.tinh_chu_vi():.2f} {doi_tuong.tinh_dien_tich():.2f}")
```

```python
#8A25
class Sach:
    def __init__(self, ten, tac_gia, nam):
        self.ten = ten
        self.tac_gia = tac_gia
        self.nam = nam
    def hien_thi(self):
        return f"{self.ten} - {self.tac_gia} - {self.nam}"

du_lieu = input().split(",")
ten = du_lieu[0].strip()
tac_gia = du_lieu[1].strip()
nam = du_lieu[2].strip()

doi_tuong = Sach(ten, tac_gia, nam)
print(doi_tuong.hien_thi())
```

```python
#8A26
class TaiKhoanNganHang:
    def __init__(self, chu_tai_khoan, so_du):
        self.chu_tai_khoan = chu_tai_khoan
        self.so_du = so_du
    def gui_tien(self, so_tien):
        self.so_du += so_tien
    def rut_tien(self, so_tien):
        if so_tien <= self.so_du:
            self.so_du -= so_tien
        else:
            print("So du khong du")
    def lay_so_du(self):
        return self.so_du

ten = input().strip()
so_du = float(input())
tien_gui = float(input())
tien_rut = float(input())

tai_khoan = TaiKhoanNganHang(ten, so_du)
tai_khoan.gui_tien(tien_gui)
tai_khoan.rut_tien(tien_rut)

print(f"{tai_khoan.lay_so_du():.2f}")
```

```python
#8A27
class SinhVien:
    def __init__(self, ten, diem):
        self.ten = ten
        self.diem = diem
    def xep_loai(self):
        if self.diem >= 8:
            return "Gioi"
        elif self.diem >= 6.5:
            return "Kha"
        elif self.diem >= 5:
            return "Trung Binh"
        else:
            return "Yeu"

ten = input().strip()
diem = float(input())

sv = SinhVien(ten, diem)
print(f"{sv.ten} - {sv.xep_loai()}")
```

```python
#8A28
class HinhTron:
    def __init__(self, ban_kinh):
        self.ban_kinh = ban_kinh
    def tinh_chu_vi(self):
        return 2 * 3.14 * self.ban_kinh
    def tinh_dien_tich(self):
        return 3.14 * (self.ban_kinh ** 2)

r = float(input())
hinh = HinhTron(r)

print(f"{hinh.tinh_chu_vi():.2f} {hinh.tinh_dien_tich():.2f}")
```

```python
#8A29
class PhanSo:
    def __init__(self, tu, mau):
        self.tu = tu
        self.mau = mau

    def cong(self, ps2):
        tu_moi = self.tu * ps2.mau + ps2.tu * self.mau
        mau_moi = self.mau * ps2.mau
        return f"{tu_moi}/{mau_moi}"

du_lieu1 = input().split("/")
du_lieu2 = input().split("/")

ps1 = PhanSo(int(du_lieu1[0]), int(du_lieu1[1]))
ps2 = PhanSo(int(du_lieu2[0]), int(du_lieu2[1]))

print(ps1.cong(ps2))
```

```python
#8A30
class DaySo:
    def __init__(self, danh_sach):
        self.danh_sach = danh_sach
    def tinh_tong(self):
        return sum(self.danh_sach)

du_lieu = list(map(int, input().split()))
doi_tuong = DaySo(du_lieu)

print(doi_tuong.tinh_tong())
```
